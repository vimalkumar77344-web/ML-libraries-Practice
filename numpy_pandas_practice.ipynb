```python
import numpy as np
# Defining a standard Python list
my_list=[10,20,30,40,50]
# Converting the Python list into a 1D NumPy array for faster computations
my_array=np.array(my_list)
print("my frist array:",my_array)
print("the type of the array:",type(my_array))
double_array=my_array*2
added_array=my_array+50
# Vectorized operation: Multiplying each element in the array by 2
print("double array =",double_array)
# Vectorized operation: Adding 50 to each element in the array
print("added array =",added_array)
```

```python
import numpy as np
# Creating a 2D NumPy array (matrix) with 2 rows and 3 columns
my_matrix=np.array([[1,2,3,],[4,5,6,]])
print("my matrix:\n",my_matrix)
# Checking the dimensions (rows, columns) of the matrix
print("\n matrix's shape(row,column)\n",my_matrix.shape)
# NumPy Broadcasting: Multiplying every element of the matrix by 10
matrix_boost=my_matrix*10
print("boost matrix is:",matrix_boost)
```

```python
#indexing & slicing
import numpy as np
#creatint a 3*3 matrix
matrix = np.array([[10,20,30],[40,50,60],[70,80,90]])
print("original matrix:\n",matrix)
print("\n single element(60):",matrix[1,2])
print("print a compete row (row 1):",matrix[1,:])
#here mean of ':' , the complete column's element
print("print a complete column (column 0):",matrix[:,0])
#here mean of ':' , the complete row's element

#slicing

sub_matrix = matrix[1:,1:]
print("sub-matrix:\n",sub_matrix)
```

```python
#mathematical & statistical functions - matrix math(axis=0,axis=1)
import numpy as np

matrix = np.array([[10,20,30,],[40,50,60]])
print("sum of complete matrix :",matrix.sum())
# Axis = 0 (Column-wise Sum) -> add Up to bottom
print("coloumn wise sum:",matrix.sum(axis=0))

# Axis = 1 (row-wise Sum) -> add left to right
print("row wise sum:",matrix.sum(axis=1))

# axis = 0 -> average of matrix according to column
print("column wise average:",matrix.mean(axis=0))

# Axis = 1 -> biggest number in all matrix (row wise)
print("biggest number in matrix(row wise):",matrix.max(axis=1))

# standard deviation
print("standard deviation:",matrix.std())
```

```python
#reshaping array

import numpy as np

flat_array = np.array([10,20,30,40,50,60])
print("original 1d shape:",flat_array.shape)
matrix_2d = flat_array.reshape(2,3)
print("\nreshaped 2d matrix:\n",matrix_2d)
print("new shape:",matrix_2d.shape,"\n")


#-1 logic in reshaping for automatic calculation

auto_matrix = flat_array.reshape(2,-1)
print("Auto-calculated Matrix (2, -1):\n", auto_matrix)
print("Auto Shape:", auto_matrix.shape)
```

```python
#Advanced Slicing & Filtering (Boolean Indexing)

import numpy as np
scores = np.array([25,55,38,89,42])

mask = scores>40
print("boolean mask:",mask)

filtered_scores = scores[mask]
print("filtered array:",filtered_scores)
```

```python
#NumPy Broadcasting (Different Size Matrix Math)

matrix = np.array([[1, 2, 3],[4, 5, 6]]) # Shape (2, 3)

row_vector = np.array([10, 20, 30]) #shape (1,3)

result = matrix + row_vector

print(result)
```

```python
#Matrix Multiplication (Dot Product)

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("using @ (A @ B):")
matrix_product = A @ B
print(matrix_product,"\n")

#instead of @ we can use np.dot(A,B)

print("using dot product (np.dot(A,B)):")
matrix_product = np.dot(A,B)
print(matrix_product)
```

```python
import pandas as pd

data = {
     'Name': ['Amit', 'Priya', 'Rahul', 'Sneha'],
    'Age':[20,30,12,23],
    'City': ['Delhi', 'Mumbai', 'Bangalore', 'Delhi'],
    'Marks': [85, 92, 78, 95]
}

df = pd.DataFrame(data)
print(df)

#only print the data of students whose city  delhi
delhi_students = df[df['City']=='Delhi']
print(delhi_students)

#only print the name and marks of students whose marks more then 80
toppers = df[df['Marks'] > 80]
print(toppers[['Name','Marks']])

#creating new columns or update old colummns

df['updated_marks'] = df['Marks']+5
print(df)

#updating some information in given data
df.loc[df['Name'] == 'Rahul', ['Marks','City']] = [59,'pune']
print(df)

#add new data

df.loc[4] = ['Vikram', 52, 'Chennai', 86 ,91]
print(df)
```

```python
import pandas as pd
import numpy as np

#incorrect data
incorrect_data = {
    'Name': ['Rahul', 'Sneha', 'Amit', 'Priya', 'Vikram'],
    'Age': [21, np.nan, 22, 20, np.nan], # np.nan matlab khali jagah
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Bangalore', 'Mumbai'],
    'Marks': [85, 90, np.nan, 78, 92]
}

df_messy = pd.DataFrame(incorrect_data)
print("Original Messy Data:\n", df_messy)
print("data shape:",df_messy.shape)

#checking the empty values in column
print(df_messy.isnull().sum())

#filling the null values
average_age = df_messy['Age'].mean()
df_messy['Age']=df_messy['Age'].fillna(average_age)
print(df_messy)

#only delete the nan values(it can be one nan in whole row ) row according to marks column
#it delete complete row(dropna)
df_final = df_messy.dropna(subset=['Marks'])
print(df_final)

#finding the small summary of given data
city_summary = df_messy.groupby('City')['Marks'].mean()
print(city_summary)
```
