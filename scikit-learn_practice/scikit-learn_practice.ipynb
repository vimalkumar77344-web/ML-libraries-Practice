```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
data = {
    'Age': [25, 30, np.nan, 22, 35],
    'Salary': [50000, np.nan, 60000, 45000, 80000]
}
df = pd.DataFrame(data)
print(df)
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
cleaned_data = imputer.fit_transform(df)
df_cleaned = pd.DataFrame(cleaned_data, columns=df.columns)
print(df_cleaned)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df = pd.DataFrame({'City': ['Delhi', 'Mumbai', 'Jaipur']})
encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(df[["City"]])
#df_encoded = pd.get_dummies(df, columns=['City'])

df_clean = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['City']))
print(df_clean)
```
