```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
data = {
    'Age': [25, 30, np.nan, 22, 35],
    'Salary': [50000, np.nan, 60000, 45000, 80000]
}
df = pd.DataFrame(data)
print(df)
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
cleaned_data = imputer.fit_transform(df)
df_cleaned = pd.DataFrame(cleaned_data, columns=df.columns)
print(df_cleaned)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df = pd.DataFrame({'City': ['Delhi', 'Mumbai', 'Jaipur']})
encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(df[["City"]])
#df_encoded = pd.get_dummies(df, columns=['City'])

df_clean = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['City']))
print(df_clean)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    'Age':[20,30,25],
    'Salary': [50000, 75000, 100000]
}
df = pd.DataFrame(data)

scaler = StandardScaler()
scaled_array = scaler.fit_transform(df)

df_scaled = pd.DataFrame(scaled_array, columns=df.columns)

print(df_scaled)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
data = {
    'Size': [
        500, 600, 700, 800, 900, 1000, 1100, 1200, 1300, 1400,
        1500, 1600, 1700, 1800, 1900, 2000, 2100, 2200, 2300, 2500
    ],
    'Price': [
        25, 31, 35, 40, 46, 52, 55, 61, 67, 72,
        76, 83, 88, 94, 99, 105, 110, 116, 121, 132
    ]
}
df = pd.DataFrame(data)
x = df[['Size']]
y = df[['Price']]
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
model = LinearRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
print("accurance of modal:",r2_score(y_test,y_pred))
print("intercep:",model.intercept_)
print("slope coefficient:",model.coef_[0])
```
