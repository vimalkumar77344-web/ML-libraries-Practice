```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
data = {
    'Age': [25, 30, np.nan, 22, 35],
    'Salary': [50000, np.nan, 60000, 45000, 80000]
}
df = pd.DataFrame(data)
print(df)
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
cleaned_data = imputer.fit_transform(df)
df_cleaned = pd.DataFrame(cleaned_data, columns=df.columns)
print(df_cleaned)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df = pd.DataFrame({'City': ['Delhi', 'Mumbai', 'Jaipur']})
encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(df[["City"]])
#df_encoded = pd.get_dummies(df, columns=['City'])

print("new columns name:",encoder.get_feature_names_out(['City']))
print(encoded_data)
```

```python
import pandas as pd
from sklearn.cluster import KMeans
data = {
    'Hours': [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 5.0, 5.5, 6.0, 6.5, 8.0, 8.5, 9.0, 9.5, 10.0],
    'Score': [10,  15,  12,  20,  25,  30,  50,  55,  48,  60,  85,  90,  88,  95,  100]
}
df = pd.DataFrame(data)
x = df[['Hours','Score']]
model = KMeans(n_clusters=3,random_state=42)
y_clusters = model.fit_predict(x)
df['cluster'] = y_clusters
print(df)
print(model.cluster_centers_)
```
